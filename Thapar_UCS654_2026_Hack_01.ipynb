```python
import numpy as np
import pandas as pd

from sklearn.model_selection import StratifiedKFold
from sklearn.impute import SimpleImputer
from sklearn.ensemble import ExtraTreesClassifier
from sklearn.metrics import accuracy_score
```

```python
train_df = pd.read_csv("train.csv")
test_df = pd.read_csv("test.csv")

print("Train shape:", train_df.shape)
print("Test shape:", test_df.shape)

ID_COL = "id"
TARGET_COL = "target"

X = train_df.drop(columns=[ID_COL, TARGET_COL])
y = train_df[TARGET_COL]

X_test = test_df.drop(columns=[ID_COL])
```

```python
imputer = SimpleImputer(strategy="median")

X = pd.DataFrame(imputer.fit_transform(X), columns=X.columns)
X_test = pd.DataFrame(imputer.transform(X_test), columns=X_test.columns)
```

```python
N_SPLITS = 10
NUM_CLASSES = y.nunique()

skf = StratifiedKFold(
    n_splits=N_SPLITS,
    shuffle=True,
    random_state=42
)

oof_preds = np.zeros((len(X), NUM_CLASSES))
test_preds = np.zeros((len(X_test), NUM_CLASSES))
```

```python
for fold, (train_idx, val_idx) in enumerate(skf.split(X, y)):
    print(f"\n Fold {fold+1}")

    X_train, X_val = X.iloc[train_idx], X.iloc[val_idx]
    y_train, y_val = y.iloc[train_idx], y.iloc[val_idx]

    model = ExtraTreesClassifier(
        n_estimators=1000,       # strong but CPU safe
        max_depth=None,
        min_samples_leaf=3,      # controls overfitting
        max_features="sqrt",
        bootstrap=False,
        n_jobs=-1,
        random_state=42
    )

    model.fit(X_train, y_train)

    oof_preds[val_idx] = model.predict_proba(X_val)
    test_preds += model.predict_proba(X_test) / N_SPLITS
```

```python
oof_labels = np.argmax(oof_preds, axis=1)
cv_acc = accuracy_score(y, oof_labels)

print("\n CV Accuracy:", round(cv_acc, 4))
```

```python
final_preds = np.argmax(test_preds, axis=1)

submission = pd.DataFrame({
    "id": test_df[ID_COL],
    "target": final_preds.astype(int)
})

submission.to_csv("submission.csv", index=False)
submission.head()
```
